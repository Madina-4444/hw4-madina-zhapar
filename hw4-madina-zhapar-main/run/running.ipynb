```python
#C1
import json
import os

from openai import OpenAI
from tools import lookup, calculate


client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"]
)

MODEL = "deepseek/deepseek-v4-flash-0731"

TOOLS = [
    {
        "type": "function",
        "function": {
            "name": "lookup",
            "description": "Looks things up.",
            "parameters": {
                "type": "object",
                "properties": {
                    "district": {"type": "string"},
                    "field": {"type": "string"}
                }
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "calculate",
            "description": "Safely evaluate arithmetic expressions.",
            "parameters": {
                "type": "object",
                "properties": {
                    "expression": {"type": "string"}
                }
            }
        }
    }
]


def load_questions():
    with open(
        "data/questions.json",
        "r",
        encoding="utf-8"
    ) as f:
        return json.load(f)


def run_question(question):

    messages = [
        {
            "role": "system",
            "content": """
Use the available tools to answer questions.

End every answer with:

ANSWER: <name or yes/no> | <number>
"""
        },
        {
            "role": "user",
            "content": question
        }
    ]

    calls = 0

    for _ in range(15):

        response = client.chat.completions.create(
            model=MODEL,
            messages=messages,
            tools=TOOLS,
            max_tokens=500
        )

        message = response.choices[0].message

        if not message.tool_calls:

            return {
                "calls": calls,
                "answer": message.content or "NO ANSWER"
            }

        messages.append(message)

        for tool_call in message.tool_calls:

            calls += 1

            try:
                args = json.loads(
                    tool_call.function.arguments
                )
            except:
                args = {}

            if tool_call.function.name == "lookup":

                if (
                    "district" not in args
                    or
                    "field" not in args
                ):
                    result = {
                        "error": "missing arguments"
                    }
                else:
                    result = lookup(
                        args["district"],
                        args["field"]
                    )

            elif tool_call.function.name == "calculate":

                if "expression" not in args:
                    result = {
                        "error": "missing arguments"
                    }
                else:
                    result = calculate(
                        args["expression"]
                    )

            else:
                result = {
                    "error": "unknown tool"
                }

            messages.append(
                {
                    "role": "tool",
                    "tool_call_id": tool_call.id,
                    "content": json.dumps(result)
                }
            )

    return {
        "calls": calls,
        "answer": "UNANSWERED"
    }


questions = load_questions()

wanted = {
    "Q-03",
    "Q-05",
    "Q-08",
    "Q-17",
    "Q-20"
}

for q in questions:

    if q["id"] not in wanted:
        continue

    result = run_question(
        q["question"]
    )

    print("\n", q["id"])
    print("Calls:", result["calls"])
    print("Answer:", result["answer"])
```
